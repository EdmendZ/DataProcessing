```python
# 1. 使用 Pandas 创建一个Series对象，数据为[10, 20, 30, 40]，索引为['a', 'b', 'c', 'd']，
# 然后获取索引为'b'的元素值。
s = pd.Series(data=[10, 20, 30, 40], index=['a', 'b', 'c', 'd'])
print(s['b'])
```

```python
# 2. 创建一个包含'姓名'和'年龄'两列的DataFrame,
# 然后计算所有人员的平均年龄。 ---获取年龄字段 df['年龄']或者df.年龄
data = {
    '姓名': ['张三', '李四', '王五'],
    '年龄': [18, 19, 20]
}
df = pd.DataFrame(data)
mean_val = df['年龄'].mean()
print(mean_val)
```

```python
import pandas as pd

data = {
    '姓名': ['Alice', 'Bob', 'Charlie', 'David'],
    '年龄': [25, 30, 35, 40],
    '城市': ['北京', '上海', '广州', '深圳']
}

# 自定义行标签为 A, B, C, D
df = pd.DataFrame(data, index=['A', 'B', 'C', 'D'])
print(df)
```

```python
print(df.loc['B', '年龄'])
print(df.loc[['A', 'C'], ['年龄', '城市']])
print(df.loc['A':'C', :])
```

```python
import pandas as pd

df1 = pd.DataFrame({
    'key': ['A', 'B', 'C'],
    'value1': [1, 2, 3]
})

df2 = pd.DataFrame({
    'key': ['B', 'C', 'D'],
    'value2': [4, 5, 6]S
})

# 合并两个 DataFrame，并处理列名冲突，如果通过on指定列，只是指定的df1。会用
# df1指定的列和df2的索引连接，会报错。除非df2.set_index指定索引。
df1.join(df2,lsuffix='_left',rsuffix='_right')
```
