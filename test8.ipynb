```python
import pandas as pd

df1 = pd.DataFrame({
    "employee": ["Bob", "Jake", "Lisa", "Sue"],
    "group": ["Accounting", "Engineering", "Engineering", "HR"]
})

df2 = pd.DataFrame({
    "name": ["Bob", "Jake", "Lisa", "Sue"],
    "salary": [70000, 80000, 120000, 90000]
})

# 内连接
df_merge_inner = pd.merge(df1, df2, left_on="employee", right_on="name", how="inner")
print(df_merge_inner)
```

```python
# 全外连接
df_merge_outer = pd.merge(df1, df2, left_on="employee", right_on="name", how="outer")
print(df_merge_outer)
```

```python
# 左外连接
df_merge_left = pd.merge(df1, df2, left_on="employee", right_on="name", how="left")
print(df_merge_left)
```

```python
# 右外连接
df_merge_right = pd.merge(df1, df2, left_on="employee", right_on="name", how="right")
print(df_merge_right)
```

```python
import pandas as pd
import numpy as np

# 1. 创建 DataFrame
data = {
    "地区": ["东部", "东部", "西部", "西部"],
    "产品": ["苹果", "香蕉", "苹果", "香蕉"],
    "销售额": [100, 150, 120, 80]
}
df = pd.DataFrame(data)

# 2. 使用 pivot_table 创建透视表
# index: 行索引
# columns: 列索引
# values: 计算的值
# aggfunc: 聚合函数（题目要求计算总和，所以用 'sum'）
pivot_df = pd.pivot_table(
    df,
    values="销售额",
    index="地区",
    columns="产品",
    aggfunc="sum"
)

print(pivot_df)
```
